## TP1 - Web scraping

Script de téléchargement automatique de fichiers PDFs depuis un site.

# Import de libraries

```python
import os
import requests
from bs4 import BeautifulSoup
```

# Récupération et téléchargement

```python
headers = {'User-Agent': 'Mozilla/5.0 (Macintosh; Intel Mac OS X 10_10_1) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/39.0.2171.95 Safari/537.36'}

filesPDF = []

root_url = f"https://max.de.wilde.web.ulb.be/camille"
response = requests.get(root_url, headers=headers)
soup = BeautifulSoup(response.text, 'html.parser')
for link in soup.find_all("a"):
    title = link.text.strip()
    url = root_url + "/" + link.get('href') if not link.get('href').startswith('http') else link.get('href')
    filesPDF.append([url, title])

print(f"La liste comprend {len(filesPDF)} fichiers.")
```

# Téléchargement des PDFs

```python
# Créer un répertoire s'il n'existe pas encore
destPath = './data'
if not os.path.exists(destPath):
    os.mkdir(destPath)
```

```python
# boucle qui télécharge les fichiers dans destPath
cpt = 0
for url, filename in filesPDF:
    response = requests.get(url)
    with open(os.path.join(destPath, filename), 'wb') as pdf_file:
        pdf_file.write(response.content)
    cpt += 1

print(f"{cpt} files copied to {destPath}")
```
