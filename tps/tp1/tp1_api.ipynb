# Récupération via API

```python
""" Code inspired from: https://www.mediawiki.org/wiki/API:Search """

url = "https://fr.wikipedia.org/w/api.php"
contentToSearch = "Le Soir"

session = requests.Session()

params = {
    "action": "query",
    "format": "json",
    "list": "search",
    "srwhat": "text",
    "srsearch": contentToSearch
}

response = session.get(url=url, params=params)
data = response.json()

if data ['query']['search'][0]['title'] == contentToSearch:
    print(f"The search page {contentToSearch} exists on French Wikipedia")
```
