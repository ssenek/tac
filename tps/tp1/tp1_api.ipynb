# Récupération via API

```python
""" Code inspired from: https://www.mediawiki.org/wiki/API:Search """

url = "https://fr.wikipedia.org/w/api.php"
contentToSearch = "Le Soir"

session = requests.Session()

headers = {
    "User-Agent": "MyPythonScript/1.0 (sasa.senek@ulb.be) requests"
}

params = {
    "action": "query",
    "format": "json",
    # Adding a "generator" for getting data about a set of pages (https://www.mediawiki.org/wiki/API:Query#Generators)
    # This helps me querying and working with several pages containing the searched content.
    "generator": "search",
    "gsrsearch": contentToSearch,
    "gsrlimit": 1, # Taking only the top X result

    "prop": "extracts",
    "explaintext": 1, # Clean text, no HTML
    "exlimit": 1 #Process only X results
}

response = session.get(url=url, headers=headers, params=params)
data = response.json()

pages = data.get("query", {}).get("pages", {}) #{} = emtpy dictionnary to avoid AttributeError in case query returns "None"

if pages:
    print(f"Found {len(pages)} results for {contentToSearch}:\n")
    for pageInfo in pages.values():
        title = pageInfo.get("title")
        extract = pageInfo.get("extract", "Nothing found")

        print(f"--- Top match: {title} ---")
        print(extract)
```
